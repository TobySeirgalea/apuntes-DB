Lista todos los géneros, ordenados alfabéticamente.

```sql
SELECT G.Name
FROM Genre G
ORDER BY G.Name
```

Muestra nombre y duración en minutos (2 decimales) de las 10 canciones más largas.

```sql
SELECT TOP 10 S.Name, S.Milliseconds
FROM Track S
ORDER BY S.Milliseconds DESC
```

Clientes que no tienen valor en Company

```sql
SELECT C.FirstName, C.LastName, C.CustomerId
FROM CUSTOMER C
WHERE C.Company IS NULL
```

Nombre completo del cliente (CONCAT) y longitud de su email, ordenados por longitud.

```sql
SELECT CONCAT(C.FirstName, ' ', C.LastName), LEN(C.Email)
FROM Customer C
ORDER BY LEN(C.Email)
```

Cada álbum con el nombre de su artista.

```sql
SELECT A.Title, Ar.Name
FROM Album A 
    INNER JOIN Artist Ar
        ON A.ArtistId = Ar.ArtistId
```

Cada canción con álbum, artista, género y tipo de medio.

```sql
SELECT T.Name, A.Title, Ar.Name, G.Name, M.Name
FROM Track T
    INNER JOIN Album A
        ON A.AlbumId = T.AlbumId
    INNER JOIN Artist Ar
        ON Ar.ArtistId = A.ArtistId
    INNER JOIN Genre G
        ON G.GenreId = T.GenreId
    INNER JOIN MediaType M
        ON T.MediaTypeId = M.MediaTypeId
```

Todos los artistas, incluso los que no tienen álbumes, con su cantidad de álbumes (LEFT JOIN).

```sql
SELECT Ar.ArtistId, COUNT(A.AlbumId)
FROM Artist Ar
    LEFT JOIN Album A
        ON Ar.ArtistId = A.ArtistId
GROUP BY Ar.ArtistId
```

Artistas sin ningún álbum.

```sql
SELECT R.ArtistId, R.Name
FROM Artist R
WHERE R.ArtistId NOT IN (SELECT A.ArtistId FROM Album A) -- No hace falta el JOIN por ArtistID porque Album ya contiene ese atributo
```

Canciones que no están en ninguna playlist.

```sql
SELECT T.TrackId, T.Name
FROM Track T
WHERE NOT EXISTS (SELECT P.TrackId FROM PlaylistTrack P WHERE P.TrackId = T.TrackId)
```

Canciones nunca vendidas.

```sql
SELECT T.TrackId, T.Name
FROM Track T
WHERE NOT EXISTS (SELECT 1 FROM InvoiceLine I WHERE I.TrackId = T.TrackId)
```

Cada empleado con el nombre de su jefe, incluyendo al que no tiene

```sql
SELECT E.FirstName, E.LastName, E.ReportsTo, B.FirstName, B.LastName
FROM Employee E
    LEFT JOIN Employee B
        ON B.EmployeeId = E.ReportsTo
```
